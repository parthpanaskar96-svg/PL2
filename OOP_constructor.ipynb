#PYTHON

```
class Student:
    def __init__(self, name, roll_no, marks):
        # Constructor initializes instance variables
        self.name = name
        self.roll_no = roll_no
        self.marks = marks
        print(f"Student object created for: {self.name}")

    def display(self):
        print(f"Name: {self.name}, Roll No: {self.roll_no}, Marks: {self.marks}")

# Constructor is called automatically when object is created
s1 = Student("Alice", 101, 95)
s1.display()

s2 = Student("Bob", 102, 88)
s2.display()

```
