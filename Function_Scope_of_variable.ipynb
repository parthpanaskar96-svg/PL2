```
# Demonstrate scope of variables
x = 'global'          # Global variable

def outer():
    x = 'enclosing'   # Enclosing scope

    def inner():
        x = 'local'   # Local scope
        print('inner()  :', x)

    inner()
    print('outer()  :', x)

def modify_global():
    global x
    x = 'modified global'

print('Before   :', x)
outer()
print('After outer():', x)
modify_global()
print('After modify():', x)












```
