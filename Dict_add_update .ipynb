#PYTHON

```
# Add and update dictionary elements
student = {'name': 'student', 'marks': 75}

# Add new key
student['roll_no'] = 103
print('After adding roll_no:', student)

# Update existing key
student['marks'] = 88
print('After updating marks:', student)

# update() method
student.update({'branch': 'Mechanical', 'email': 'student@gmail.com'})
print('After update():', student)

```
