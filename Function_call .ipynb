```
# Define and call a function
def greet(name):
    print(f'Hello, {name}! Welcome to Python.')

def add(a, b):
    print(f'Sum of {a} and {b} is {a + b}')

# Calling the functions
greet('Parth')
greet('Shree')
add(10, 20)
add(5, 7)





```
