```
# Sort dictionary by keys and values
students = {
    'Shree': 82,
    'Aditya'  : 95,
    'Parth'    : 78,
    'Dishant'    : 91,
    'Kunal'   : 88
}

# Sort by keys (ascending)
sorted_by_key = dict(sorted(students.items()))
print('Sorted by Key (A-Z):')
for k, v in sorted_by_key.items():
    print(f'  {k}: {v}')

# Sort by values (descending)
sorted_by_val = dict(sorted(students.items(), key=lambda x: x[1], reverse=True))
print('\nSorted by Marks (High to Low):')
for k, v in sorted_by_val.items():
    print(f'  {k}: {v}')




```
