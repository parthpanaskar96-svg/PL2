```
# Calculate factorial using recursion
def factorial_recursive(n):
    if n < 0:
        return 'Not defined'
    if n == 0 or n == 1:   # Base case
        return 1
    return n * factorial_recursive(n - 1)  # Recursive call

# Test the function
for num in [0, 1, 5, 7, 10]:
    print(f'{num}! = {factorial_recursive(num)}')









```
