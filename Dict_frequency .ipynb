```
# Count frequency of words in a sentence using dictionary
sentence = input('Enter a sentence: ')
words = sentence.lower().split()

freq = {}
for word in words:
    freq[word] = freq.get(word, 0) + 1

print('\nWord Frequency:')
for word, count in sorted(freq.items()):
    print(f'  {word}: {count}')


```
