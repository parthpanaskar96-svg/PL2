# Python
Day converter into year , month and weeks


```
day=int(input("Enter the number of days:"))
year=day/365
month=day/30
week=day/7
print(f"{day} equals to {year:.4f} years")
print(f"{day} equals to {month:.4f} months")
print(f"{day} equals to {week:.4f} weeks")
```
