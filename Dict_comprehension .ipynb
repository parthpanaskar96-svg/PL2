```
# Dictionary comprehension examples

# 1. Squares of numbers 1-10
squares = {x: x**2 for x in range(1, 11)}
print('Squares:', squares)

# 2. Filter students with marks >= 85
students = {'Aditya': 95, 'Shree': 78, 'Dishant': 91, 'Kunal': 82, 'Parth': 88}
passed = {name: marks for name, marks in students.items() if marks >= 85}
print('\nStudents with marks >= 85:', passed)

# 3. Swap keys and values
swapped = {v: k for k, v in students.items()}
print('\nSwapped (marks -> name):', swapped)

# 4. Word lengths
words = ['python', 'dictionary', 'comprehension']
lengths = {word: len(word) for word in words}
print('\nWord lengths:', lengths)




```
