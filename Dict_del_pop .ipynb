```
# Delete elements from a dictionary
student = {
    'name': 'Student77', 'roll_no': 545,
    'branch': 'IT', 'marks': 80, 'email': 'student@gmail.com'
}

# del keyword
del student['email']
print('After del email:', student)

# pop() method
removed = student.pop('branch')
print('Popped value:', removed)
print('After pop:', student)

# popitem() - removes last inserted
last = student.popitem()
print('Popped item:', last)
print('After popitem:', student)

# clear()
student.clear()
print('After clear():', student)


```
