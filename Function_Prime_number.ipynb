```
# Check prime number using a function
def is_prime(n):
    if n < 2:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False
    for i in range(3, int(n**0.5) + 1, 2):
        if n % i == 0:
            return False
    return True

# Test individual numbers
test_nums = [1, 2, 3, 4, 17, 25, 29, 49, 97]
for num in test_nums:
    status = 'Prime' if is_prime(num) else 'Not Prime'
    print(f'{num:3d} -> {status}')

# Print all primes up to 50
primes = [n for n in range(2, 51) if is_prime(n)]
print('\nPrimes up to 50:', primes)












```
