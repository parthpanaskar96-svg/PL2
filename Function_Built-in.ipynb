```
# Demonstrate Python built-in functions
nums = [4, 7, 1, 9, 3, 6, 2, 8, 5]
text = 'Hello, Python!'

# Numeric built-ins
print('List      :', nums)
print('len()     :', len(nums))
print('sum()     :', sum(nums))
print('min()     :', min(nums))
print('max()     :', max(nums))
print('sorted()  :', sorted(nums))
print('abs(-7)   :', abs(-7))
print('round(3.7):', round(3.7))
print('pow(2,10) :', pow(2, 10))

# String / type built-ins
print('type(nums):', type(nums))
print('len(text) :', len(text))
print('upper()   :', text.upper())
print('range(5)  :', list(range(5)))










```
