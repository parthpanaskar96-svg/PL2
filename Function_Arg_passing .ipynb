```
# Passing arguments to a function

# Positional arguments
def student_info(name, roll, branch):
    print(f'Name: {name}, Roll: {roll}, Branch: {branch}')

# Default arguments
def power(base, exp=2):
    print(f'{base}^{exp} = {base ** exp}')

# Keyword arguments
def display(city, country='India'):
    print(f'City: {city}, Country: {country}')

student_info('Alice', 101, 'CS')
student_info(roll=102, name='Bhau', branch='IT')
power(3)
power(2, 8)
display('Pune')
display(city='Delhi', country='India')






```
