```
# Access keys, values, and items of a dictionary
student = {
    'name': 'Admin',
    'roll_no': 102,
    'branch': 'Electronics',
    'marks': 87.0
}

print('Keys:')
print(list(student.keys()))

print('\nValues:')
print(list(student.values()))

print('\nItems:')
for key, value in student.items():
    print(f'  {key}: {value}')


```
