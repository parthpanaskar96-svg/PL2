```
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner           # Public attribute
        self._account_type = "Savings"  # Protected attribute
        self.__balance = balance     # Private attribute (name mangled)

    # Getter method
    def get_balance(self):
        return self.__balance

    # Setter method with validation
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: {amount}. New Balance: {self.__balance}")
        else:
            print("Invalid deposit amount!")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrawn: {amount}. New Balance: {self.__balance}")
        else:
            print("Insufficient funds or invalid amount!")

    def display(self):
        print(f"Owner: {self.owner}")
        print(f"Account Type: {self._account_type}")
        print(f"Balance: {self.__balance}")

acc = BankAccount("Alice", 10000)
acc.display()

acc.deposit(5000)
acc.withdraw(3000)
acc.withdraw(20000)  # Should fail

# Accessing private attribute directly would raise AttributeError
# print(acc.__balance)  # This would fail
print(f"\nBalance via getter: {acc.get_balance()}")

```
