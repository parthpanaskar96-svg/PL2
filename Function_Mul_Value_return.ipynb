```
# Return multiple values from a function
def arithmetic(a, b):
    add = a + b
    sub = a - b
    mul = a * b
    div = a / b if b != 0 else 'undefined'
    return add, sub, mul, div

def min_max(numbers):
    return min(numbers), max(numbers)

# Unpack returned values
a, b = 15, 4
plus, minus, times, divide = arithmetic(a, b)
print(f'Add: {plus}, Sub: {minus}, Mul: {times}, Div: {divide}')

nums = [3, 7, 1, 9, 4]
low, high = min_max(nums)
print(f'Min: {low}, Max: {high}')







```
