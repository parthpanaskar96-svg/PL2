#PYTHON

```
class GrandParent:
    def grand_method(self):
        print("This is GrandParent class method")

class Parent(GrandParent):
    def parent_method(self):
        print("This is Parent class method")

# Child inherits from Parent which inherits from GrandParent
class Child(Parent):
    def child_method(self):
        print("This is Child class method")

c = Child()
c.grand_method()   # From GrandParent
c.parent_method()  # From Parent
c.child_method()   # Own method

print("\nInheritance Chain:")
print("GrandParent -> Parent -> Child")
print("MRO:", [cls.__name__ for cls in Child.__mro__])
```
