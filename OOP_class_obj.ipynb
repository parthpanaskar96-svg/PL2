#PYTHON

```
class Dog:
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed

    def display(self):
        print(f"Dog Name: {self.name}, Breed: {self.breed}")

# Creating an object
dog1 = Dog("Buddy", "Labrador")
dog1.display()

dog2 = Dog("Max", "German Shepherd")
dog2.display()
```
